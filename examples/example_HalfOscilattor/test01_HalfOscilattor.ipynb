```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import hermite
from math import factorial

# --- 1. FUNÇÃO TEÓRICA (Mantida) ---
def theoretical_half_oscillator(k, x):
    n_full = 2 * k + 1
    norm_const = 1.0 / np.sqrt(2**n_full * factorial(n_full) * np.sqrt(np.pi))
    H_n = hermite(n_full)
    psi = np.sqrt(2) * norm_const * H_n(x) * np.exp(-0.5 * x**2)
    return psi

# --- 2. CONFIGURAÇÃO ---
arquivos = ['HalfOscilattor/HalfOscilattor_psi_level_0.txt', 'HalfOscilattor/HalfOscilattor_psi_level_1.txt', 'HalfOscilattor/HalfOscilattor_psi_level_2.txt'] # Seus arquivos
niveis = [0, 1, 2]
cores = ['blue', 'green', 'red']

# LISTA DE MARCADORES DIFERENTES
# 'o' = Círculo
# 's' = Quadrado (Square)
# '^' = Triângulo para cima
marcadores = ['o', 's', '^'] 

plt.figure(figsize=(10, 7))

# --- 3. LOOP DE PLOTAGEM ---
# Adicionamos 'marcadores' ao zip para pegar um diferente a cada volta
for k, arquivo, cor, marker_style in zip(niveis, arquivos, cores, marcadores):
    
    # A) Teoria
    x_teorico = np.linspace(0, 9.128187521926685, 1000) 
    psi_teorico = theoretical_half_oscillator(k, x_teorico)
    
    # B) Numérico
    try:
        dados = np.loadtxt(arquivo) 
        x_num = dados[:, 0]
        psi_num = dados[:, 1]
        
        # Correção de sinal (se necessário)
        idx_max = np.argmax(np.abs(psi_num))
        if np.sign(psi_num[idx_max]) != np.sign(theoretical_half_oscillator(k, x_num[idx_max])):
            psi_num = psi_num * -1

        # --- PLOT DA TEORIA (Linha sólida) ---
        plt.plot(x_teorico, psi_teorico, 
                 color=cor, linestyle='-', alpha=1, linewidth=1.5,
                 label=f'Theory (n={k})')
        
        # --- PLOT DO NUMÉRICO (Figuras Ocas) ---
        plt.plot(x_num, psi_num, 
                 marker=marker_style,       # Usa o marcador da lista (o, s, ^)
                 linestyle='',              # Remove a linha conectando os pontos
                 color=cor,                 # Define a cor base
                 markerfacecolor='none',    # DEIXA O CENTRO TRANSPARENTE (OCO)
                 markeredgecolor=cor,       # Pinta a borda com a cor do nível
                 markeredgewidth=0.7,       # Engrossa um pouco a borda para ver melhor
                 markersize=7,              # Tamanho do marcador
                 markevery=15,               # Ajuste se tiver pontos demais (ex: a cada 5)
                 label=f'PySpectral (n={k})')
                 
    except Exception as e:
        print(f"Erro no arquivo {arquivo}: {e}")

# --- 4. ESTÉTICA ---
plt.title("Comparação: PySpectral vs Solução Analítica (Meio Oscilador)")
plt.xlabel("x")
plt.ylabel("$\psi_n(x)$")
plt.axhline(0, color='black', linewidth=0.8) # Linha zero
plt.legend()
plt.grid(True, alpha=0.3)
plt.xlim(0, 9.128187521926685) # Ajuste conforme seus dados

plt.show()
```
