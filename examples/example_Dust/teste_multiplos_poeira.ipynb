```python
from PySpectral import *
```

```python
# superposição equiprovável para os 3 primeiros níveis:

niveis = 100
superposicao = 3

coeficientes = [1]*superposicao + [0]*(niveis - superposicao)


# intervalo de tempo considerado:
# [0, 100]
tempo = 100

# total de pontos calculados dentro do intervalo:
pontos = 300
```

```python
import numpy as np

# Lista de combinações de parâmetros que você quer testar
param_sets = [
    {"k": 1, "l": -1, "b": 0.27, "s": 0.9999},
    {"k": 2, "l": -1, "b": 0.5,  "s": 1.0},
    {"k": 1, "l": -2, "b": 0.3,  "s": 0.95},
]

for params in param_sets:
    k = params["k"]
    l = params["l"]
    b = params["b"]
    s = params["s"]

    # Calcula grandezas derivadas
    m = (6*(b+2))/s
    omega = ((2*k)/(b+2))**0.5

    # Define funções do problema
    def f(x, m=m, omega=omega, l=l):
        return (m**2)*(omega**2) - 2*l*m*(x**4)

    def g(x, m=m):
        return 2*m*x

    def w(x):
        return x

    # Cria root_filename dinâmico com base nos parâmetros
    root_filename = f"k{k}_l{l}_b{b}_s{s}"

    # Instancia o problema
    p = SpectralMethod(
        num_levels=niveis,
        length=10,
        f_function=f,
        g_function=g,
        weight=w,
        num_digits=15,
        root_filename=root_filename,
        label=root_filename,
        optimizer_L='y'
    )

    print(f"\n  Performing analysis for {root_filename}")
    p.run_full_analysis(coeficientes=coeficientes, pontos=pontos, tempo=tempo)
```
